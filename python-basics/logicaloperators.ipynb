```python
## logical operators
a = True
b = False

## and
c = a and b
print(c)

## or
d = a or b
print(d)

## not
e = not a
print(e)
```

```python
## Simple Calculator 

a = float(input("Enter a number: "))
b = float(input("Enter another number: "))

## Addition
c = a + b
print(c)

## Subtraction
d = a - b
print(d)

## Multiplication
e = a * b
print(e)

## Division
f = a / b
print(f)
```
