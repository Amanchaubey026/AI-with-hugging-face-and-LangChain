```python
## elif is used when there are multiple conditions to check
a = 15
if a > 20:
    print("a is greater than 20")
elif a > 10:
    print("a is greater than 10")
else:
    print("a is not greater than 10")
```

```python
## nested conditional statement
men = "brave"
if men == "bravee":
    print("men are bravee")
else:
    if men == "braveee":
        print("men are braveee")
    elif men == "brave":
        print("men are brave")
    else:
        print("men are not brave")
```

```python
## determine if the year is a leap year or not 
year  = float(input("Enter the year:"))
if ((year % 400 == 0) or ( year %100 != 0) and (year % 4 == 0)):
    print("This is a leap year")
else:
    print("This is not a leap year")
```

```python
## simple calculator
num1 = float(input("Enter first number:"))
num2 = float(input("Enter the second number:"))
operator = input("Enter the operator +,-,*,/,//,% :")

if operator == "+":
    print("Sum is : ", num1 + num2)
elif operator == "-":
    print("Difference is : ", num1 - num2)
elif operator == "*":
    print("Product id : ", num1 * num2)
elif operator == "/":
    print("Division is :", num1 / num2)
elif operator == "//":
    print("Floor Division is :", num1 // num2)
elif operator == "%":
    print("Remainder is :", num1 % num2)
else:
    print("Invalid operator")
```
